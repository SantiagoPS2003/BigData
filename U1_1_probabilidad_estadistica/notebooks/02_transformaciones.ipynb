```python
import pandas as pd
import requests

# Fetch the data.
df = pd.read_csv("https://ourworldindata.org/grapher/life-expectancy.csv?v=1&csvType=full&useColumnShortNames=true", storage_options = {'User-Agent': 'Our World In Data data fetch/1.0'})

# Fetch the metadata
metadata = requests.get("https://ourworldindata.org/grapher/life-expectancy.metadata.json?v=1&csvType=full&useColumnShortNames=true").json()
```

```python
df.head()
```

```python
df_1970 = df[(df["year"] == 1970) & (df["entity"].str.contains("^K", regex=True))]
df_1970
```

```python
df.set_index("entity")
```

```python
df.set_index(["code","year"])
```

```python
df.head(3)
```

```python
df.rename(columns={"entity" : "pais"},inplace=True)
```

```python
df.loc[df["code"].isnull()]
```

```python
df.loc[df["code"].notnull(), ["pais","year","life_expectancy_0"]].head(3)
```

```python
df.dropna(subset=["code"], inplace=True)
df.isnull().sum()
```

```python
resultado = (
    df.groupby("pais")["life_expectancy_0"]
    .mean()
    .sort_values(ascending=True)
)
resultado.head(10)
```

```python
new_df = df[df["year"].between(1970 , 2023)]

resultado = (
    new_df.groupby("pais")["life_expectancy_0"]
    .mean()
    .sort_values(ascending=True)
)
resultado.head(10)
```
